```python
for col in numeric_cols:

    plt.figure(figsize=(8,3))

    plt.boxplot(df[col].dropna(),vert=False)

    plt.title(col)

    plt.show()
```

```python
outlier_summary = []

for col in numeric_cols:

    Q1 = df[col].quantile(.25)

    Q3 = df[col].quantile(.75)

    IQR = Q3-Q1

    lower = Q1-1.5*IQR

    upper = Q3+1.5*IQR

    outliers = ((df[col]<lower)|(df[col]>upper)).sum()

    outlier_summary.append([col,outliers])

outlier_df = pd.DataFrame(outlier_summary,

                          columns=["Column","Outliers"])

outlier_df.sort_values("Outliers",ascending=False)
```

```python
for col in numeric_cols:

    negative = (df[col]<0).sum()

    if negative>0:

        print(col,negative)
```

```python
if {'salary_min_lpa','salary_midpoint_lpa','salary_max_lpa'}.issubset(df.columns):

    wrong_salary = df[
        (df['salary_min_lpa'] > df['salary_midpoint_lpa']) |
        (df['salary_midpoint_lpa'] > df['salary_max_lpa'])
    ]

    print("Incorrect Salary Rows :",len(wrong_salary))
```

```python
if "company_rating" in df.columns:

    print(df["company_rating"].describe())
```

```python
if "company_rating" in df.columns:

    invalid = df[
        (df["company_rating"]<0) |
        (df["company_rating"]>5)
    ]

    print(len(invalid))
```

```python
print(df.memory_usage(deep=True).sum()/1024**2,"MB")
```

```python
for col in cat_cols:

    if df[col].nunique()/len(df)<0.50:

        df[col]=df[col].astype("category")
```
