```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv("cleaned_indian_tech_jobs.csv")

df.head()
```

```python
df["salary_category"] = pd.cut(
    df["salary_midpoint_lpa"],
    bins=[0,5,10,20,50,100],
    labels=[
        "Low",
        "Medium",
        "High",
        "Very High",
        "Premium"
    ]
)

df["salary_category"].value_counts()
```

```python
df["experience_group"] = pd.cut(
    df["experience_min_yrs"],
    bins=[-1,1,3,6,10,50],
    labels=[
        "Fresher",
        "Junior",
        "Mid-Level",
        "Senior",
        "Expert"
    ]
)

df["experience_group"].value_counts()
```

```python
df["rating_category"] = pd.cut(
    df["company_rating"],
    bins=[0,2,3,4,5],
    labels=[
        "Poor",
        "Average",
        "Good",
        "Excellent"
    ]
)

df["rating_category"].value_counts()
```

```python
median_salary = df["salary_midpoint_lpa"].median()

df["high_paying_job"] = np.where(
    df["salary_midpoint_lpa"] >= median_salary,
    "Yes",
    "No"
)

df["high_paying_job"].value_counts()
```

```python
df["remote_flag"] = np.where(
    df["work_mode"]=="Remote",
    1,
    0
)

df["remote_flag"].value_counts()
```

```python
df["hybrid_flag"] = np.where(
    df["work_mode"]=="Hybrid",
    1,
    0
)
```

```python
df["onsite_flag"] = np.where(
    df["work_mode"]=="On-site",
    1,
    0
)
```

```python
top10 = df["primary_city"].value_counts().head(10).index

df["top_city"] = np.where(
    df["primary_city"].isin(top10),
    "Yes",
    "No"
)

df["top_city"].value_counts()
```

```python
df["salary_range"] = (
    df["salary_max_lpa"] -
    df["salary_min_lpa"]
)

df["salary_range"].describe()
```

```python
df["avg_experience"] = (
    df["experience_min_yrs"] +
    df["experience_max_yrs"]
)/2

df["avg_experience"].head()
```

```python
df["salary_per_experience"] = (
    df["salary_midpoint_lpa"] /
    (df["avg_experience"] + 1)
)

df["salary_per_experience"].head()
```

```python
df["top_rated_company"] = np.where(
    df["company_rating"]>=4.5,
    "Yes",
    "No"
)

df["top_rated_company"].value_counts()
```

```python
df["fresher_job"] = np.where(
    df["experience_min_yrs"]==0,
    "Yes",
    "No"
)

df["fresher_job"].value_counts()
```

```python
if "posting_age_days" in df.columns:

    df["posting_age_group"] = pd.cut(
        df["posting_age_days"],
        bins=[0,7,15,30,365],
        labels=[
            "1 Week",
            "2 Weeks",
            "1 Month",
            "Old"
        ]
    )
```

```python
df.to_csv(
    "final_jobs_dataset.csv",
    index=False
)

print("Feature Engineering Completed Successfully!")
```
