```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings("ignore")

# Display Settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.float_format", lambda x: "%.2f" % x)

# Plot Settings
plt.rcParams["figure.figsize"] = (10,6)
sns.set_style("whitegrid")
```

```python
df = pd.read_csv("cleaned_indian_tech_jobs.csv")

print("Dataset Loaded Successfully!")
```

```python
plt.figure(figsize=(10,6))

sns.histplot(
    data=df,
    x="salary_midpoint_lpa",
    bins=30,
    kde=True
)

plt.title("Distribution of Salary Midpoint (LPA)")
plt.xlabel("Salary (LPA)")
plt.ylabel("Number of Jobs")

plt.show()
```

```python
plt.figure(figsize=(10,3))

sns.boxplot(
    x=df["salary_midpoint_lpa"]
)

plt.title("Salary Midpoint Distribution")

plt.show()
```

```python
if "company_rating" in df.columns:

    plt.figure(figsize=(8,5))

    sns.histplot(
        df["company_rating"],
        bins=20,
        kde=True
    )

    plt.title("Company Rating Distribution")

    plt.xlabel("Rating")

    plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.histplot(
    df["experience_min_yrs"],
    bins=15,
    kde=True
)

plt.title("Minimum Experience Required")

plt.xlabel("Experience (Years)")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="work_mode",
    order=df["work_mode"].value_counts().index
)

plt.title("Jobs by Work Mode")

plt.xlabel("Work Mode")

plt.ylabel("Number of Jobs")

plt.show()
```

```python
work_mode_percent = round(
    df["work_mode"].value_counts(normalize=True) * 100,
    2
)

work_mode_percent
```

```python
plt.figure(figsize=(7,7))

plt.pie(
    df["work_mode"].value_counts(),
    labels=df["work_mode"].value_counts().index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Work Mode Distribution")

plt.show()
```

```python
plt.figure(figsize=(9,5))

sns.countplot(
    data=df,
    x="salary_tier",
    order=df["salary_tier"].value_counts().index
)

plt.xticks(rotation=20)

plt.title("Salary Tier Distribution")

plt.show()
```

```python
plt.figure(figsize=(9,5))

sns.countplot(
    data=df,
    x="experience_tier",
    order=df["experience_tier"].value_counts().index
)

plt.title("Experience Tier Distribution")

plt.show()
```

```python
top_cities = (
    df["primary_city"]
    .value_counts()
    .head(10)
)

plt.figure(figsize=(10,6))

sns.barplot(
    x=top_cities.values,
    y=top_cities.index
)

plt.title("Top 10 Hiring Cities")

plt.xlabel("Number of Jobs")

plt.ylabel("City")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.boxplot(
    data=df,
    x="work_mode",
    y="salary_midpoint_lpa"
)

plt.title("Salary Distribution by Work Mode")
plt.xlabel("Work Mode")
plt.ylabel("Salary (LPA)")

plt.show()
```

```python
avg_salary_workmode = (
    df.groupby("work_mode")["salary_midpoint_lpa"]
      .mean()
      .sort_values(ascending=False)
      .reset_index()
)

avg_salary_workmode
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=avg_salary_workmode,
    x="work_mode",
    y="salary_midpoint_lpa"
)

plt.title("Average Salary by Work Mode")
plt.xlabel("Work Mode")
plt.ylabel("Average Salary (LPA)")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df,
    x="experience_min_yrs",
    y="salary_midpoint_lpa",
    alpha=0.6
)

plt.title("Salary vs Minimum Experience")
plt.xlabel("Experience (Years)")
plt.ylabel("Salary (LPA)")

plt.show()
```

```python
salary_exp = (
    df.groupby("experience_tier")["salary_midpoint_lpa"]
      .mean()
      .sort_values(ascending=False)
      .reset_index()
)

salary_exp
```

```python
plt.figure(figsize=(10,5))

sns.barplot(
    data=salary_exp,
    x="experience_tier",
    y="salary_midpoint_lpa"
)

plt.title("Average Salary by Experience Tier")
plt.xlabel("Experience Tier")
plt.ylabel("Average Salary (LPA)")

plt.xticks(rotation=20)

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df,
    x="company_rating",
    y="salary_midpoint_lpa",
    alpha=0.5
)

plt.title("Company Rating vs Salary")
plt.xlabel("Company Rating")
plt.ylabel("Salary (LPA)")

plt.show()
```

```python
top_companies = (
    df["company_name"]
    .value_counts()
    .head(10)
)

plt.figure(figsize=(12,6))

sns.barplot(
    x=top_companies.values,
    y=top_companies.index
)

plt.title("Top 10 Hiring Companies")
plt.xlabel("Number of Jobs")
plt.ylabel("Company")

plt.show()
```

```python
city_salary = (
    df.groupby("primary_city")["salary_midpoint_lpa"]
      .mean()
      .sort_values(ascending=False)
      .head(10)
)

city_salary
```

```python
plt.figure(figsize=(10,6))

sns.barplot(
    x=city_salary.values,
    y=city_salary.index
)

plt.title("Top 10 Highest Paying Cities")
plt.xlabel("Average Salary (LPA)")
plt.ylabel("City")

plt.show()
```

```python
highest_company_salary = (
    df.groupby("company_name")["salary_midpoint_lpa"]
      .mean()
      .sort_values(ascending=False)
      .head(10)
)

highest_company_salary
```

```python
plt.figure(figsize=(12,6))

sns.barplot(
    x=highest_company_salary.values,
    y=highest_company_salary.index
)

plt.title("Top 10 Highest Paying Companies")
plt.xlabel("Average Salary (LPA)")
plt.ylabel("Company")

plt.show()
```

```python
numeric_df = df.select_dtypes(include=["number"])

corr = numeric_df.corr()

plt.figure(figsize=(12,8))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
top_roles = df["job_title"].value_counts().head(10)

plt.figure(figsize=(12,6))

sns.barplot(
    x=top_roles.values,
    y=top_roles.index
)

plt.title("Top 10 Job Roles")
plt.xlabel("Number of Jobs")
plt.ylabel("Job Role")

plt.show()
```

```python
top_domains = df["skill_domain"].value_counts()

plt.figure(figsize=(10,6))

sns.barplot(
    x=top_domains.values,
    y=top_domains.index
)

plt.title("Job Distribution by Skill Domain")
plt.xlabel("Number of Jobs")
plt.ylabel("Skill Domain")

plt.show()
```

```python
fresher = df["is_fresher_friendly"].value_counts()

plt.figure(figsize=(6,6))

plt.pie(
    fresher.values,
    labels=fresher.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Fresher-Friendly Jobs")

plt.show()
```

```python
domain_salary = (
    df.groupby("skill_domain")["salary_midpoint_lpa"]
      .mean()
      .sort_values(ascending=False)
      .reset_index()
)

plt.figure(figsize=(10,6))

sns.barplot(
    data=domain_salary,
    x="salary_midpoint_lpa",
    y="skill_domain"
)

plt.title("Average Salary by Skill Domain")
plt.xlabel("Salary (LPA)")
plt.ylabel("Skill Domain")

plt.show()
```

```python
role_salary = (
    df.groupby("job_title")["salary_midpoint_lpa"]
      .mean()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(12,6))

sns.barplot(
    x=role_salary.values,
    y=role_salary.index
)

plt.title("Top 10 Highest Paying Job Roles")
plt.xlabel("Average Salary (LPA)")
plt.ylabel("Job Role")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="company_rating"
)

plt.title("Company Rating Distribution")

plt.show()
```
